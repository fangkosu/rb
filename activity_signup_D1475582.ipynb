```python
# ==========================================
# HW09｜兩場活動報名整理器
# 姓名：蘇芳可
# 學號：D1475582
# 測試一：原始資料
# ==========================================

# 1. 原始資料設定
activity_names = ("桌遊體驗", "電影欣賞")
board_raw = " Amy, Ben, amy, Cora "
movie_raw = " ben, Dora, ERIC, ben "
unit_fees = {"桌遊體驗": 120, "電影欣賞": 80}

# 讀取活動名稱（tuple 拆包）
board_name, movie_name = activity_names

# 整理文字：移除半形空白 -> 轉小寫 -> 依逗號拆成 list（保留重複與順序）
board_list = board_raw.replace(" ", "").lower().split(",")
movie_list = movie_raw.replace(" ", "").lower().split(",")

# 2. 轉為 set 去重，並進行集合運算
board_set = set(board_list)
movie_set = set(movie_list)

both_set = board_set & movie_set  # 共同參加者（交集）
all_set = board_set | movie_set  # 全部參加者（聯集）
only_board_set = board_set - movie_set  # 只參加桌遊者（差集）
only_movie_set = movie_set - board_set  # 只參加電影者（差集）

# 3. 排序名單與切片取得通知預覽前兩位
board_sorted = sorted(board_set)
movie_sorted = sorted(movie_set)
both_sorted = sorted(both_set)
only_board_sorted = sorted(only_board_set)
only_movie_sorted = sorted(only_movie_set)
all_sorted = sorted(all_set)

board_preview = board_sorted[:2]
movie_preview = movie_sorted[:2]

# 4. 建立統計與費用 summary 字典（動態計算）
summary = {
    "board_unique_count": len(board_set),
    "movie_unique_count": len(movie_set),
    "total_unique_count": len(all_set),
    "total_registrations": len(board_set) + len(movie_set),
    "board_revenue": len(board_set) * unit_fees[board_name],
    "movie_revenue": len(movie_set) * unit_fees[movie_name],
}
summary["total_revenue"] = summary["board_revenue"] + summary["movie_revenue"]

# 5. 輸出整理摘要
print(f"=== 【測試一】{board_name} & {movie_name} 報名整理摘要 ===")
print(f"整理後原始串列（{board_name}）：{board_list}")
print(f"整理後原始串列（{movie_name}）：{movie_list}")
print("-" * 45)
print(
    f"{board_name} 去重名單（{summary['board_unique_count']} 人）：{board_sorted}"
)
print(
    f"{movie_name} 去重名單（{summary['movie_unique_count']} 人）：{movie_sorted}"
)
print(f"兩場共同參加者：{both_sorted}")
print(f"只參加 {board_name}：{only_board_sorted}")
print(f"只參加 {movie_name}：{only_movie_sorted}")
print(f"總不重複名單（{summary['total_unique_count']} 人）：{all_sorted}")
print("-" * 45)
print(f"{board_name} 通知預覽前兩位：{board_preview}")
print(f"{movie_name} 通知預覽前兩位：{movie_preview}")
print("-" * 45)
print(f"總不重複人數：{summary['total_unique_count']} 人")
print(f"總報名人次：{summary['total_registrations']} 人次")
print(f"{board_name} 應收費用：{summary['board_revenue']:.2f} 元")
print(f"{movie_name} 應收費用：{summary['movie_revenue']:.2f} 元")
print(f"兩場總應收金額：{summary['total_revenue']:.2f} 元")
```
